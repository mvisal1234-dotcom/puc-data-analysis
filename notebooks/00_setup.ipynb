```python
from pathlib import Path

REQUIRED_FOLDERS = [
    "notebooks",
    "data/raw",
    "data/processed",
    "figures"
]

def create_project_folders(folders):
    report = {}

    for folder in folders:
        path = Path(folder)
        existed = path.exists()

        path.mkdir(parents=True, exist_ok=True)

        report[folder] = "already existed" if existed else "created"

    return report

result = create_project_folders(REQUIRED_FOLDERS)

for folder, status in result.items():
    print(f"{folder:<20} {status}")
    
```

```python
import sys
import importlib

PACKAGES = ["numpy", "pandas", "matplotlib", "seaborn", "scipy"]

def environment_report(packages=PACKAGES):
    package_versions = {}

    for name in packages:
        try:
            module = importlib.import_module(name)
            package_versions[name] = getattr(
                module, "__version__", "unknown"
            )
        except ImportError:
            package_versions[name] = None

    missing = sorted(
        name for name, version in package_versions.items()
        if version is None
    )

    return {
        "python_version": sys.version.split()[0],
        "packages": package_versions,
        "all_installed": len(missing) == 0,
        "missing": missing
    }

report = environment_report()

print("Python version:", report["python_version"])
print("All installed:", report["all_installed"])
print("Missing:", report["missing"])

print("\nPackage versions:")
for name, version in report["packages"].items():
    print(f"{name}: {version}")
```

```python
report = environment_report()

assert isinstance(report, dict)
assert set(report.keys()) == {
    "python_version",
    "packages",
    "all_installed",
    "missing"
}

assert report["all_installed"] is True
assert report["missing"] == []

probe = environment_report([
    "numpy",
    "definitely_not_a_real_package_xyz"
])

assert probe["all_installed"] is False
assert probe["missing"] == [
    "definitely_not_a_real_package_xyz"
]

print("Problem 1: all assertions passed.")
```

```python
def safe_divide(a, b):
    try:
        a = float(a)
        b = float(b)
        return a / b
    except (ZeroDivisionError, TypeError, ValueError):
        return None

for a, b in [
    (10, 2),
    (7, 0),
    ("8", "4"),
    ("abc", 2),
    (None, 5)
]:
    print(f"safe_divide({a!r}, {b!r}) = {safe_divide(a, b)}")
```

```python
assert safe_divide(10, 2) == 5.0
assert safe_divide(7, 2) == 3.5
assert safe_divide(-9, 3) == -3.0
assert safe_divide(0, 5) == 0.0

assert safe_divide("8", "4") == 2.0
assert safe_divide("3.5", "0.5") == 7.0

assert safe_divide(5, 0) is None
assert safe_divide(5, "0") is None
assert safe_divide("abc", 2) is None
assert safe_divide(None, 5) is None
assert safe_divide(10, None) is None
assert safe_divide([1, 2], 3) is None

assert isinstance(safe_divide(10, 2), float)

print("Problem 2: all assertions passed.")
```
